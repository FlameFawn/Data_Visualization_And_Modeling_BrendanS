```python
import pandas as pd
import plotly.express as px


df = pd.read_csv('Covid_States.csv')

df.head()
```

# Stock prices over time

```python
# Create an animated scatter plot
# Ensure 'positive' contains no NaN values (Plotly requires numeric sizes)
df_plot = df.copy()
df_plot['positive'] = df_plot['positive'].fillna(0)

fig = px.scatter(df_plot,
                 x="totalTestResults",
                 y="negative",
                 size="positive",
                 color="state",
                 hover_name="state",
                 log_x=True,
                 size_max=60,
                 animation_frame="date",
                 title="COVID tests over time by state",
                 labels={"totalTestResults": "Total Test Results", "negative": "Negative Tests"}
                 )

# Adjust the height of the figure and set y-axis range
fig.update_layout(height=500, yaxis_range=[20,85])  

# Display the plot
fig.show()
```
